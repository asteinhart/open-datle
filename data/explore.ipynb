```python
import polars as pl
import altair as alt
```

```python
evictions = pl.read_parquet("datasets/evictions.parquet")
```

```python
## quick analysis, evictions per year
evictions = evictions.with_columns(
    pl.col("Executed Date").str.to_date("%m/%d/%Y").alias("date_parsed")
).with_columns(
    pl.col("date_parsed").dt.year().alias("year")
)
```

```python
#evictions per year
by_year = evictions.group_by("year").count().sort("year")
by_year.plot.line(x="year", y="count")
# export as json
by_year.write_json("chart_data/evictions_per_year.json")
```

```python
# avg evictions per month over years
evictions = evictions.with_columns(
    pl.col("date_parsed").dt.month().alias("month")
)
by_month = evictions.group_by("month").count().sort("month")
by_month.plot.bar(x="month", y="count") 
```

```python
#evictions per year and month, maybe one date column for time series
by_year_month = evictions.group_by(["year", "month"]).count().sort(["year", "month"])
by_year_month = by_year_month.with_columns(
    pl.date(pl.col("year"), pl.col("month"), 1).alias("date")
)
by_year_month.plot.line(x="date", y="count")
```
